```python
import pandas as pd 
```

```python
url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"
```

```python
df = pd.read_csv(url)
df.head()
```

### Check for Missing Values

We check each column for missing values before making any changes.

- `df.isna()` identifies missing values.
- `.sum()` counts the missing values in each column.

This helps us decide which columns need cleaning. This step does not change the data.

```python
df.isna().sum()
```

### Findings: Missing Values

Four columns contain missing values:

- `age`: 177 missing values.
- `embarked`: 2 missing values.
- `deck`: 688 missing values, the highest count.
- `embark_town`: 2 missing values.

All other columns have no missing values detected by `isna()`.
No data has been changed yet.

```python
df.shape
```

### Calculate the Percentage of Missing Values

The dataset has 891 rows and 15 columns.
We divide each column's missing count by the total number of rows
and multiply by 100 to calculate its missing percentage.

```python
missing_percentage = (df.isna().sum() / len(df)) * 100 # i calculate ang percentage sa missing value
missing_percentage.round(2)
```

### Finding: Missing Deck Information

The `deck` column has 688 missing values out of 891 records (77.22%).
Only 203 records have deck information.
We need to consider this limitation before using this column in analysis.

### Inspect Data Types

We use `df.info()` to inspect the columns, non-missing counts,
and data types. This helps us check whether each column uses
an appropriate data type before cleaning.

```python
df.info()
```

### Findings: Data Types

The dataset contains integer, float, string, and boolean columns.

### Inspect Numeric Values

We use descriptive statistics to review numeric columns.
Minimum and maximum values help us identify values that may need investigation.

### Findings: Numeric Values

Recorded ages range from 0.42 to 80 years, with no negative ages.

### Findings: Zero Fares

There are 15 records with a fare of zero across different ticket classes.
The available data does not explain why these fares are zero.
We retain these records because there is insufficient evidence to treat them as errors.

### Check for Possible Duplicate Rows

We count rows that repeat the same values across all columns.
Matching rows require review because different passengers may share the same recorded details.

```python
df.duplicated().sum()
```

### Review Matching Rows

We found 107 repeated rows, excluding the first occurrence in each group.
We will inspect matching records before deciding whether to remove any.
The dataset has no passenger ID or name to confirm passenger identity.

```python
df[df.duplicated(keep=False)].head(10)
```

```python
df.loc[[32,47]]
```

### Decision: Retain Matching Records

Rows 32 and 47 have identical recorded values, including missing age and deck.
Without passenger identifiers, we cannot confirm that matching rows represent the same person.
We retain all records to avoid removing potentially different passengers.

### Create a Working Copy

We create a separate copy for cleaning.
The original data remains in `df`, while cleaning changes will be made in `df_clean`.

```python
df_clean = df.copy()
df.shape
```

```python
df_clean["deck"] = df_clean["deck"].fillna("Unknown")

df_clean["deck"].value_counts()
```

### Label Missing Embarkation Towns

We replace missing embarkation towns with "Unknown".
This keeps all records without guessing where these passengers boarded.

```python
df_clean["embark_town"] = df_clean["embark_town"].fillna("Unknown")

df_clean["embark_town"].value_counts()
```

### Inspect Missing Embarkation Codes

We inspect records with missing `embarked` values and compare their
`embark_town` values before deciding how to handle them.

```python
df_clean.loc[
    df_clean["embarked"].isna(),
    ["embarked", "embark_town"]

    ]
```

### Label Missing Embarkation Codes

Rows 61 and 829 have missing embarkation codes and unknown embarkation towns.
We label the missing codes as "Unknown" rather than guessing a boarding location.

```python
df_clean["embarked"] = df_clean["embarked"].fillna("Unknown")

df_clean.loc[[61,829], ["embarked","embark_town"]]
```

### Preserve and Flag Missing Ages

We keep missing ages as NaN to avoid inventing passenger ages.
We add an `missing_age` column: True means the age is missing,
and False means an age is recorded.

```python
df_clean["missing_age"] = df_clean["age"].isna()

df_clean["missing_age"].value_counts()
```

### Validate Missing Values and Dataset Size

We check the remaining missing values and confirm that all 891 records remain.
Only `age` should contain missing values.
The added `missing_age` column identifies these records.

```python
print("Dataset Shape:", df_clean.shape)
df_clean.isna().sum()
```

### Export the Cleaned Dataset

We save 891 records and 16 columns to `titanic_cleaned.csv`.
Unknown deck and embarkation information is labeled "Unknown".
Missing ages remain blank and are identified by `missing_age`.

```python
df_clean.to_csv("titanic_cleaned.csv", index=False)

saved_df = pd.read_csv("titanic_cleaned.csv")
saved_df.shape
```

### Cleaning Summary

- Preserved all 891 passenger records.
- Labeled 688 missing deck values as "Unknown".
- Labeled 2 missing values each in embarked and embark_town as "Unknown".
- Kept 177 missing ages and added the missing_age indicator.
- Retained zero fares and matching rows because they were not confirmed errors.
- Exported the dataset with 16 columns to titanic_cleaned.csv.

### Limitations

Unknown information remains unknown after labeling.
Age-based analysis uses only recorded ages unless an explicit
imputation method is applied. Matching records cannot be confirmed
as duplicate passengers without identifiers.
